```python
import cmath
import matplotlib.pyplot as plt
import numpy as np
```

The mandelbrot set is given by the iterative formula:
$$z_{n+1} = z_{n}^{2}+z_{n}$$

```python
z0 = [complex(0.05, 0.02)]
n=1000000
for i in range(n):
    z1 = z0[i]
    z2 = (z1)**2 + (z1)
    z0.append(z2)

plt.scatter([i.real for i in z0], [i.imag for i in z0], s=5)
plt.xlabel("Re(z)")
plt.ylabel("Im(z)")
plt.title("Mandelbrot transformations of complex numbers")
plt.grid(alpha=0.1)
```
