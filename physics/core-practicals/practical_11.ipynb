To save time, I just copied and pasted the excel data rather than replotting them in Python.

```python
import pandas as pd
from io import StringIO

raw = """
Angle(degrees)   1    2    3    Induced-V
0                6.2  5.8  5.9  0.298333333
10               5.8  5.6  5.7  0.285
20               5.8  5.4  5.5  0.278333333
30               5.4  4.8  4.9  0.251666667
40               4.6  4.4  4.3  0.221666667
50               3.8  3.6  3.5  0.181666667
60               3.2  2.8  2.7  0.145
70               2.2  2.0  1.9  0.101666667
80               1.2  0.6  0.7  0.041666667
90               0    0    0    0
"""

df = pd.read_csv(StringIO(raw), sep=r"\s+")
print(df)
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
k = (df['Induced-V'])
```

I then replotted the graph to ensure that the results are still consistent, with an overlay

```python
plt.scatter(df['Angle(degrees)'], df['Induced-V'], s=10, c='k', label='Discrete Data (Induced Voltage)')
x = np.linspace(0, np.pi/2, 10)
q = np.linspace(0, 90, 10)
plt.plot(q, k.max() * np.cos(x), label='cos(x)')
plt.grid(alpha=0.23)
plt.xlim(0, 90)
plt.legend()
plt.title('Comparison of collected data to cos(x)')
plt.xlabel('Angle (degrees)')
plt.ylabel('Induced Voltage (V)')
```

I then came across the Pearson Correlation Coefficient (PCC) as a way of quantifying the relationship between my data and $\cos{x}$ without just giving a qualitative comparison.

```python
corr = np.corrcoef(k, k.max() * np.cos(x))[0, 1]
print("Similarity to cos(x):", corr)
```

As values of the PCC tend towards $1$, it is more likely that the results are almost identical (a complete correlation) and since - especially with the demonstrated error bars - there seems to be a very strong correlation I can conclude that this experiment has verified accurately the relationship between induced emf ($\epsilon$) and angle.
